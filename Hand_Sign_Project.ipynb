# Building Model Using CNN

```python
class CustomCNN(nn.Module):
    def __init__(self, num_classes=28):
        super(CustomCNN, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2,2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2,2),

            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),
            nn.MaxPool2d(2,2),

            nn.Conv2d(256, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(),
            nn.Conv2d(512, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(),
            nn.MaxPool2d(2,2),
        )
        self.classifier = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

model = CustomCNN(num_classes=28).to(device)
```

# Loss and Optimization

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

# Train Model

```python
def train_model(model, train_data_load, criterion, optimizer):
    model.train()
    running_loss = 0.0

    for imgs, labels in train_data_load:
        imgs, labels = imgs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    return running_loss
```

```python
def validate_model(model, val_data_load, criterion):
    model.eval()
    val_loss = 0.0
    correct = 0 # Counts correct predictions
    total = 0 # counts total images

    with torch.no_grad():
        for imgs, labels in val_data_load:
            imgs, labels = imgs.to(device), labels.to(device)
            outputs = model(imgs)
            loss = criterion(outputs, labels)

            val_loss += loss.item()

            _, preds = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (preds == labels).sum().item()

    val_accuracy = 100 * correct / total
    return val_loss, val_accuracy
```

```python
epochs = 25
for epoch in range(epochs):
    train_loss = train_model(model, train_data_load, criterion, optimizer)
    val_loss, val_accuracy = validate_model(model, val_data_load, criterion)

    print(f"Epoch {epoch+1}/{epochs}, Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}, Val Accuracy: {val_accuracy:.2f}%")
```

```python
# Save the model after training
torch.save(model.state_dict(), '/content/drive/MyDrive/ASL_Image_models/cnn_custom_model_1.pth')
```

# Evaluation Metrics

```python
from sklearn.metrics import classification_report

# No need of device code at all
model.eval()

all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in val_data_load:
        # Move the input images and labels to the same device as the model
        images, labels = images.to(device), labels.to(device)

        # Forward pass
        outputs = model(images)  # directly images
        _, preds = torch.max(outputs, 1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())  # .cpu() to bring back to CPU for sklearn

# Now calculate metrics
print(classification_report(all_labels, all_preds, digits=4))
```

```python
%%writefile app.py
import streamlit as st
import torch
import torch.nn as nn
from torchvision import models, transforms
from PIL import Image
import numpy as np

# Load the model
model = models.resnet18()
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 28)  # 28 letters + space + nothing
model.load_state_dict(torch.load('/content/drive/MyDrive/ASL_Image_models/ASL_resnet_model.pth', map_location=torch.device('cpu')))
model.eval()

# Image preprocessing
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

# ASL Labels (your order)
asl_labels = [
    'A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J', 'K', 'L', 'M', 'N',
    'Nothing', 'O', 'P', 'Q', 'R', 'S', 'Space', 'T', 'U', 'V', 'W', 'X', 'Y', 'Z'
]

st.set_page_config(page_title="ASL Recognition", layout="wide")
st.title("ASL Alphabet Recognition")

# Prediction function
def predict_asl(image_file):
    img = Image.open(image_file).convert('RGB')
    img = transform(img).unsqueeze(0)
    with torch.no_grad():
        outputs = model(img)
        probabilities = torch.softmax(outputs, dim=1)
        confidence, preds = torch.max(probabilities, 1)
    predicted_label = asl_labels[preds.item()]

    # Handle 'Space'
    if predicted_label.lower() == 'space':
        predicted_label = ' '
    return predicted_label, confidence.item()

# Home page
def home_page():
    st.write("### ASL (American Sign Language) is a visual language that uses hand gestures to represent letters of the alphabet.")
    st.write("This app helps in recognizing individual ASL alphabet signs using a machine learning model trained on ASL hand gesture images.")
    st.write("### Instructions:")
    st.write("1. Go to the 'Upload Image' tab in the navigation bar.")
    st.write("2. Upload images of your hand gestures (single or multiple images).")
    st.write("3. The app will predict the corresponding letter(s) and display the final word.")

# Upload page
def upload_page():
    uploaded_files = st.file_uploader("Choose images...", type=["jpg", "jpeg", "png"], accept_multiple_files=True)
    if uploaded_files:
        word = ""
        for uploaded_file in uploaded_files:
            img = Image.open(uploaded_file).convert('RGB')
            st.image(img, caption='Uploaded Image', use_container_width=False, width=200)
            predicted_letter = predict_asl(uploaded_file)
            if predicted_letter:
              if isinstance(predicted_letter, tuple):
                    predicted_letter = predicted_letter[0]
              st.write(f"Predicted Letter: {predicted_letter.strip()}")
              word += predicted_letter
        st.write(f"### Final Predicted Word: `{word.strip()}`")

# Sidebar navigation
st.sidebar.title("Navigation")
page = st.sidebar.radio("Select a page:", ["Home", "Upload Image"])

if page == "Home":
    home_page()
else:
    upload_page()



```
